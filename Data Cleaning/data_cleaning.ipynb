```python
import re

def extract_age(age):
    age_num = re.findall('[0-9]+', str(age))
    if len(age_num) > 0:
        return age_num[0]
    else:
        return age
df['Age'] = df['Age'].apply(extract_age) #Extract age values from the 'Age' column.
```

```python
df_age = df[df['Age'] != 'nan years']['Age'] #Filter the DataFrame to exclude rows where the 'Age' column contains the string 'nan years'.
```

```python
age_median = int(df_age.dropna().astype('int64').median())   #Calculate the median of the 'Age' column after dropping missing values and converting the data type to integer.
```

```python
age_median
```

```python
df.replace('nan years', age_median, inplace=True) #Replace occurrences of the string 'nan years' in the DataFrame with the calculated median age.   
```

```python
df['Age'].fillna(age_median, inplace=True) #Fill missing values in the 'Age' column with the mean of the column.
```

```python
df['Age'].unique() #Return the unique values in the 'Age' column.
```

```python
df['Purchase_Amount'].fillna(df['Purchase_Amount'].median(), inplace=True) #Fill missing values in the 'Purchase_Amount' column with the median of the column.
```

```python
df['Feedback_Score'].fillna(df['Feedback_Score'].mode()[0], inplace=True) #Fill missing values in the 'Feedback' column with the mode of the column.
```

```python
df.isnull().sum() #Check for missing values in the DataFrame and return the count of missing values for each column.
```

```python
for col in ['Gender' , 'City' , 'Country']:
    df[col].fillna(df[col].mode()[0], inplace=True) #Fill missing values in the specified columns with the mode of each column.
```

```python
df['Last_Purchase_Date'].ffill(inplace=True) #Fill missing values in the 'Last_Purchase_Date' column using forward fill method.
```
